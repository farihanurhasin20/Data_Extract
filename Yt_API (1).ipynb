```python
from googleapiclient.discovery import build
import pandas as pd
import time

# Replace with your YouTube API Key
API_KEY = "AIzaSyDaU0OP38XkFvUv7YfE_779kjK0A2Cl2Jo"
CHANNEL_ID = "UC8XI9zhD5YgMh75y-R_HFBQ"

# Initialize YouTube API
youtube = build("youtube", "v3", developerKey=API_KEY)

# Step 1: Get the Uploads Playlist ID (where all channel videos are stored)
def get_uploads_playlist_id(channel_id):
    request = youtube.channels().list(
        part="contentDetails",
        id=channel_id
    )
    response = request.execute()
    return response["items"][0]["contentDetails"]["relatedPlaylists"]["uploads"]

# Step 2: Fetch All Video IDs from the Playlist
def get_all_video_ids(playlist_id):
    video_ids = []
    request = youtube.playlistItems().list(
        part="snippet",
        playlistId=playlist_id,
        maxResults=50  # Fetch up to 50 videos per request
    )

    while request:
        response = request.execute()
        for item in response.get("items", []):
            video_ids.append(item["snippet"]["resourceId"]["videoId"])

        # Handle pagination
        next_page_token = response.get("nextPageToken")
        if not next_page_token:
            break

        time.sleep(1)  # Prevent hitting API limits
        request = youtube.playlistItems().list(
            part="snippet",
            playlistId=playlist_id,
            maxResults=50,
            pageToken=next_page_token
        )

    return video_ids

# Step 3: Fetch All Comments & Replies for a Given Video ID
def get_comments_with_replies(video_id):
    comments = []
    request = youtube.commentThreads().list(
        part="snippet,replies",
        videoId=video_id,
        maxResults=100
    )

    while request:
        response = request.execute()
        for item in response.get("items", []):
            top_comment = item["snippet"]["topLevelComment"]["snippet"]
            comments.append([
                top_comment["authorDisplayName"],  # Author
                top_comment["publishedAt"],       # Timestamp
                top_comment["likeCount"],         # Like count
                top_comment["textDisplay"],       # Comment text
                video_id,                         # Video ID
                "Top-level"                       # Comment type
            ])

            # Check for replies
            if "replies" in item:
                for reply in item["replies"]["comments"]:
                    reply_snippet = reply["snippet"]
                    comments.append([
                        reply_snippet["authorDisplayName"],  # Reply author
                        reply_snippet["publishedAt"],       # Timestamp
                        reply_snippet["likeCount"],         # Like count
                        reply_snippet["textDisplay"],       # Reply text
                        video_id,                           # Video ID
                        "Reply"                             # Comment type
                    ])

        # Handle pagination
        next_page_token = response.get("nextPageToken")
        if not next_page_token:
            break

        time.sleep(1)
        request = youtube.commentThreads().list(
            part="snippet,replies",
            videoId=video_id,
            maxResults=100,
            pageToken=next_page_token
        )

    return comments

# Step 4: Fetch Comments & Replies for All Videos in the Channel
def get_all_channel_comments(channel_id):
    # Get the uploads playlist ID
    uploads_playlist_id = get_uploads_playlist_id(channel_id)
    print("Uploads Playlist ID:", uploads_playlist_id)

    # Get all video IDs
    video_ids = get_all_video_ids(uploads_playlist_id)
    print("Total Videos Found:", len(video_ids))

    # Fetch comments & replies for each video
    all_comments = []
    for video_id in video_ids:
        print(f"Fetching comments for Video ID: {video_id}")
        comments = get_comments_with_replies(video_id)
        all_comments.extend(comments)

    return all_comments

# Run the script
print(f"Fetching all comments & replies for channel ID: {CHANNEL_ID}")
comments_data = get_all_channel_comments(CHANNEL_ID)

# Save to CSV
df = pd.DataFrame(comments_data, columns=["Author", "Published At", "Like Count", "Comment", "Video ID", "Type"])
df.to_csv("youtube_channel_comments_with_replies.csv", index=False)
print("All comments & replies saved to youtube_channel_comments_with_replies.csv")
```
